```python
# --- 1. Load Human Interactome --- \nhuman_graph = download_string_db_data(organism_id='9606', score_threshold=700)
```

```python
# --- 2. Define Terminals --- \nsource_node = '9606.ENSP00000275493' # EGFR\nsink_node = '9606.ENSP00000371303'   # MYC\n\nif source_node not in human_graph or sink_node not in human_graph:\n    print('Warning: One or both terminal nodes not found in the graph.')\n    all_nodes = list(human_graph.nodes())\n    source_node = all_nodes[0]\n    sink_node = all_nodes[-1]\n    print(f'Using fallback terminals: {source_node}, {sink_node}')
```
